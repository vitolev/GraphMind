```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
gcn_raw = pd.read_csv("hyperparameter_search_results_gcn.csv")
```

```python
# Rows with same model,learning_rate,epochs,hidden_dim,num_layers,dropout columns values group together
gcn_grouped = gcn_raw.groupby(['model', 'learning_rate', 'epochs', 'hidden_dim', 'num_layers', 'dropout']).mean().reset_index()
# Number of rows
print(f"Number of unique hyperparameter combinations: {gcn_grouped.shape[0]}")
```

```python
# Sort by val_loss and take top 10
top_10 = gcn_grouped.nsmallest(10, 'val_loss')
print("Top 10 hyperparameter combinations by validation loss:")
display(top_10)
```
